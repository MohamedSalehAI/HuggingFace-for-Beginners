## Change Event

```python
with gr.Blocks() as app:
    with gr.Row():
        x_slider = gr.Slider()
        y_slider = gr.Slider()

    with gr.Row():
        result = gr.Text()

    y_slider.change(fn=multiply,
                    inputs=[x_slider, y_slider],
                    outputs=result)

app.launch(share=True)
```

```python
with gr.Blocks() as demo:
    with gr.Row():
        x_slider = gr.Slider(label="X")
        y_slider = gr.Slider(label="Y")

    with gr.Row():
        result = gr.Text()

    x_slider.change(fn=multiply,
                    inputs=[x_slider, y_slider],
                    outputs=result)

    y_slider.change(fn=multiply,
                    inputs=[x_slider, y_slider],
                    outputs=result)

demo.launch(share=True)
```

## Click event (Button Click triggers code)

```python
with gr.Blocks() as demo:
    with gr.Row():
        x_slider = gr.Slider(label="X",value="val_x")
        y_slider = gr.Slider(label="Y",value="val_y")

    with gr.Row():
        result = gr.Text()

    with gr.Row():
        button = gr.Button(value="Multiply!")

    button.click(fn=multiply,
                 inputs=[x_slider, y_slider],
                 outputs=[result])

demo.launch(share=True)
```

## Returning different outputs

```python
def make_grayscale(image_path):
    # Open the image

    image = Image.open(image_path)

    # Convert the image to grayscale format
    image_grayscale = image.convert("L")

    return image_grayscale, "Image converted"
```

```python
gray_image, log = make_grayscale("image_3.jpg")
```

## Gradio integration with machine learning

```python
from transformers import AutoImageProcessor, AutoModelForImageClassification

processor = AutoImageProcessor.from_pretrained("microsoft/resnet-18")
model = AutoModelForImageClassification.from_pretrained("microsoft/resnet-18")
```

```python
model
```

```python
model.config.id2label
```

```python
def classify_image(image):

    # Processing image
    image = processor(image, return_tensors="pt")["pixel_values"]

    # Passing the image to the model -> logits 0-1000
    logits = model(image).logits

    # Grab the index of the highest probability
    predicted_label = logits.argmax(-1).item()

    return model.config.id2label[predicted_label]
```

```python
iface = gr.Interface(fn=classify_image,
                     inputs=gr.Image(),
                     outputs=gr.Label(),
                     description="Upload an image")

iface.launch()
```

## Gradio Errors

```python
def validate_even(num):
    return num % 2 == 0
```

```python
validate_even(12)
```

```python
def add_even(a, b):
    if not validate_even(a):
        gr.Info("FIRST VALUE IS NOT EVEN!")

    if not validate_even(b):
        gr.Info("SECOND VALUE IS NOT EVEN!")

    return a + b
```

```python
interface = gr.Interface(fn=add_even,
                     inputs=[gr.Number(), gr.Number()],
                     outputs=gr.Number())

interface.launch()
```

#Chatbot Example

```python
import gradio as gr

def chatbot(message, history):
    reply = f"You said: {message}"
    history.append((message, reply))
    return "", history

with gr.Blocks() as demo:
    chatbot_ui = gr.Chatbot()
    msg = gr.Textbox(label="Message")
    clear = gr.Button("Clear")

    msg.submit(
        chatbot,
        inputs=[msg, chatbot_ui],
        outputs=[msg, chatbot_ui]
    )

    clear.click(lambda: [], outputs=chatbot_ui)

demo.launch()
```

#Image + Prompt

```python
import gradio as gr
from PIL import ImageDraw

def process(image, prompt):
    image = image.copy()
    draw = ImageDraw.Draw(image)
    draw.text((20, 20), prompt, fill="red")
    return image

with gr.Blocks() as demo:
    gr.Markdown("# AI Image Demo")

    with gr.Row():
        image = gr.Image(type="pil")
        prompt = gr.Textbox(label="Prompt")

    output = gr.Image()

    generate = gr.Button("Generate")

    generate.click(
        fn=process,
        inputs=[image, prompt],
        outputs=output
    )

demo.launch()
```
