#different ways of sampling generation a text

```python
def logits(sentence):
    token_ids = tokenizer(sentence, return_tensors="pt").input_ids
    logits = model(token_ids).logits
    return logits[0,-1]
logits=logits("Once upon a time, in a small village,")

def greedy_decode(logits):
    """Return token index with maximum probability."""
    return torch.argmax(logits, dim=-1)

# TOP K SAMPLING
print("greedy_decode",greedy_decode(outputs),":",tokenizer.decode(greedy_decode(outputs)))

def top_k_sampling(logits, k=50):
    """
    keeps only top-k logits, normalize them into probability.
    them sample one token from the filtered distribution.
    """
    values, indices = torch.topk(logits, k)
    probs = torch.softmax(values, dim=-1)
    sampled = torch.multinomial(probs, 10)
    return indices[sampled]
print("top_k_sampling",top_k_sampling(outputs),":",tokenizer.decode(top_k_sampling(outputs)))

# Top-p (Nuecles) Sampling

def top_p_sampling(logits, p=0.9):
    """
    Sort tokens by probability, keep smallest number whose culumative
    probability exceeds threshold p, then sample one token.
    """

    sorted_logits, sorted_indices = torch.sort(logits, descending=True)
    sorted_probs = torch.softmax(sorted_logits, dim=-1)
    cumulative_probs = sorted_probs.cumsum(dim=-1)

    # Mask token outside nuclues
    mask = cumulative_probs > p
    # print("mask",mask)
    sorted_logits[mask] = float("-inf")
    # print("logits_mask",sorted_logits)

    # Sample from filtered logits
    filtered_probs = torch.softmax(sorted_logits, dim=-1)
    sampled = torch.multinomial(filtered_probs, 10)

    # Return token index in originial vocabulary
    return sorted_indices[sampled]

print("top_p_sampling",top_p_sampling(outputs, p=0.9),":",tokenizer.decode(top_p_sampling(outputs, p=0.9)))

## Temperature Sampling ##
def temperature_sampling(logits, temperature=1.0):
    """
    Scale logits by temperature before sampling.
    Lower temperature => sharper distribution
    """

    scaled = logits / temperature
    probs = torch.softmax(scaled, dim=-1)
    return torch.multinomial(probs, 10)
print("temperature_sampling",temperature_sampling(logits),tokenizer.decode(temperature_sampling(logits)))

## Random Sampling ##

def random_sampling(logits):
    """
    Sample dirctly from softmax distribution without filtring
    """

    probs = torch.softmax(logits, dim=-1)
    return torch.multinomial(probs, 10)

print("random_sampling",random_sampling(logits),":",tokenizer.decode(random_sampling(logits)))


```

## Question Answering

```python
from transformers import AutoTokenizer, AutoModelForQuestionAnswering
import torch

model_name = "distilbert-base-cased-distilled-squad"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForQuestionAnswering.from_pretrained(model_name)

context = "Paris is the capital of France."
question = "What is the capital of France?"
def qa_bot(question, context):
  inputs = tokenizer(question, context, return_tensors="pt")
  outputs = model(**inputs)

  answer_start = torch.argmax(outputs.start_logits)
  answer_end = torch.argmax(outputs.end_logits) + 1

answer = tokenizer.decode(inputs["input_ids"][0][answer_start:answer_end])
print(answer)
```

## Machine Translation

```python
import transformers
print(transformers.__version__)
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "Helsinki-NLP/opus-mt-en-fr"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)
def translator(sentence):
  encoded_input = tokenizer(sentence,return_tensors='pt')
  output=model.generate(**encoded_input)
  return tokenizer.decode(output[0], skip_special_tokens=True)
```

```python
print(translator("Hello"))
```

```python
translator("Thanks")
```

```python
sentence = "What is your name?"
translator(sentence)
```
